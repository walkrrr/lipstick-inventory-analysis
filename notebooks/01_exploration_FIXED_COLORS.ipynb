# Lipstick Inventory – Exploratory Data Analysis

This notebook explores a personal lipstick inventory using Python and pandas.
The goal is to understand brand distribution, color undertones, and how well
the collection aligns with a Soft / True Autumn color palette.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(
    'https://raw.githubusercontent.com/walkrrr/lipstick-inventory-analysis/main/data/makeup_inventory.csv',
    sep='\t'
)
df.head()
```

```python
# ---- Visualization Palette Fix (RUN THIS CELL FIRST) ----
import matplotlib.pyplot as plt

# Normalize undertone values so palette matches consistently
df['undertone'] = (
    df['undertone']
    .astype(str)
    .str.lower()
    .str.strip()
)

# Undertone color palette
undertone_palette = {
    'warm': '#C68642',
    'neutral-warm': '#E6B89C',
    'neutral': '#D9B8A6',
    'neutral-cool': '#C97C9E',
    'cool': '#8B3A62'
}
```

```python
# Normalize undertone values so palette always matches
df['undertone'] = (
    df['undertone']
    .astype(str)
    .str.lower()
    .str.strip()
)

# Undertone color palette
undertone_palette = {
    'warm': '#C68642',          # caramel brown
    'neutral-warm': '#E6B89C',  # peachy nude
    'neutral': '#D9B8A6',       # beige rose
    'neutral-cool': '#C97C9E',  # dusty rose
    'cool': '#8B3A62'           # berry / burgundy
}
```

## Brand Distribution

This chart shows how lip products are distributed across brands.
It highlights which brands dominate the collection and helps identify
brand loyalty versus variety in purchasing behavior.

```python
brand_counts = df['brand'].value_counts()
plt.figure(figsize=(10,5))
brand_counts.plot(kind='bar')
plt.title('Lip Products by Brand')
plt.xlabel('Brand')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## Undertone Distribution

This chart shows the distribution of undertones across the lipstick collection.
Colors are intentionally styled to reflect makeup tones (warm browns, nudes,
and berry shades), reinforcing the relationship between data and domain context.

```python
undertone_counts = df['undertone'].value_counts()
undertone_colors = [
    undertone_palette.get(u, '#999999')
    for u in undertone_counts.index
]

plt.figure(figsize=(7,5))
undertone_counts.plot(kind='bar', color=undertone_colors)
plt.title('Undertone Distribution of Lip Products')
plt.xlabel('Undertone')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Color Season Compatibility

This chart compares lip products that are compatible with a Soft / True Autumn
palette versus those that are not. This helps assess how closely the collection
aligns with the intended color season.

```python
df['is_autumn_friendly'].value_counts().plot(
    kind='bar', figsize=(5,5), color=['#E6B89C', '#8B3A62']
)
plt.title('Autumn-Friendly vs Non-Autumn Products')
plt.xlabel('Is Autumn Friendly')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Key Takeaways

- A significant portion of the collection does not fully align with a Soft / True Autumn palette.
- Warm and neutral-warm undertones are more likely to be autumn-friendly.
- Brand popularity does not always correlate with color season compatibility.
