```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
undertone_palette = {
    "warm": "#C68642",          # caramel brown
    "neutral-warm": "#E6B89C",  # peachy nude
    "neutral": "#D9B8A6",       # beige rose
    "neutral-cool": "#C97C9E",  # dusty rose
    "cool": "#8B3A62"           # burgundy berry
}
```

```python
brand_counts = df["brand"].value_counts()

brand_counts.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Lip Products by Brand")
plt.xlabel("Brand")
plt.ylabel("Number of Products")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
undertone_counts = df["undertone"].value_counts()

colors = [undertone_palette.get(u, "#999999") for u in undertone_counts.index]

undertone_counts.plot(
    kind="bar",
    figsize=(7, 5),
    color=colors
)

plt.title("Distribution of Lip Product Undertones")
plt.xlabel("Undertone")
plt.ylabel("Number of Products")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
autumn_counts = df["is_autumn_friendly"].value_counts()

autumn_counts.plot(
    kind="bar",
    figsize=(5, 5),
    color=["#E6B89C", "#8B3A62"]  # nude vs berry
)

plt.title("Autumn-Friendly vs Non-Autumn Lip Products")
plt.xlabel("Autumn Friendly")
plt.ylabel("Number of Products")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
(df["is_autumn_friendly"].value_counts(normalize=True) * 100).plot(
    kind="bar",
    figsize=(5, 5),
    color=["#E6B89C", "#8B3A62"]
)

plt.title("Percentage of Lip Products by Autumn Compatibility")
plt.ylabel("Percentage")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
brand_autumn = (
    df.groupby(["brand", "is_autumn_friendly"])
      .size()
      .unstack(fill_value=0)
)

brand_autumn.plot(
    kind="bar",
    stacked=True,
    figsize=(12, 6),
    color=["#8B3A62", "#E6B89C"]  # not autumn / autumn
)

plt.title("Brand Breakdown by Autumn Compatibility")
plt.xlabel("Brand")
plt.ylabel("Number of Products")
plt.xticks(rotation=45, ha="right")
plt.legend(["Not Autumn-Friendly", "Autumn-Friendly"])
plt.tight_layout()
plt.show()
```

```python
undertone_autumn = (
    df.groupby(["undertone", "is_autumn_friendly"])
      .size()
      .unstack(fill_value=0)
)

colors = ["#8B3A62", "#E6B89C"]

undertone_autumn.plot(
    kind="bar",
    stacked=True,
    figsize=(8, 5),
    color=colors
)

plt.title("Undertone vs Autumn Compatibility")
plt.xlabel("Undertone")
plt.ylabel("Number of Products")
plt.xticks(rotation=45)
plt.legend(["Not Autumn-Friendly", "Autumn-Friendly"])
plt.tight_layout()
plt.show()
```
