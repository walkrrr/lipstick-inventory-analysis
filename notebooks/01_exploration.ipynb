# Lipstick Inventory – Exploratory Data Analysis

This notebook explores a personal lipstick inventory dataset.

```
import pandas as pd
import matplotlib.pyplot as plt
```

```
df = pd.read_csv(
    'https://raw.githubusercontent.com/walkrrr/lipstick-inventory-analysis/main/data/makeup_inventory.csv',
    sep='\t'
)
df.head()
```

```
df.info()
```

## Brand Distribution

```
brand_counts = df['brand'].value_counts()
plt.figure(figsize=(10,5))
brand_counts.plot(kind='bar')
plt.title('Lip Products by Brand')
plt.xlabel('Brand')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## Undertone Distribution

```
undertone_palette = {
    'warm': '#C68642',
    'neutral-warm': '#E6B89C',
    'neutral': '#D9B8A6',
    'neutral-cool': '#C97C9E',
    'cool': '#8B3A62'
}

undertone_counts = df['undertone'].value_counts()
colors = [undertone_palette.get(u, '#999999') for u in undertone_counts.index]

plt.figure(figsize=(7,5))
undertone_counts.plot(kind='bar', color=colors)
plt.title('Undertone Distribution of Lip Products')
plt.xlabel('Undertone')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Autumn-Friendly vs Non-Autumn

```
df['is_autumn_friendly'].value_counts().plot(
    kind='bar', figsize=(5,5), color=['#E6B89C', '#8B3A62']
)
plt.title('Autumn-Friendly vs Non-Autumn Products')
plt.xlabel('Is Autumn Friendly')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
